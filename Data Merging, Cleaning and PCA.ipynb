```python
import pandas as pd

# Define paths
file_path = r"D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Final_Cleaned_Dataset_for_Stats.csv"
output_path = r"D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Filtered_ML_Ready_Dataset.csv"

# Load the cleaned 838-genome dataset
df = pd.read_csv(file_path)

# Separate metadata and features
metadata = df[['Isolate ID', 'MLST']]
features = df.drop(columns=['Isolate ID', 'MLST'])

# 1. Calculate prevalence (proportion of presence) for each gene per ST
prevalence_by_st = features.groupby(metadata['MLST']).mean()

# 2. Apply Condition (i): > 40% in at least one ST
cond_i = (prevalence_by_st > 0.40).any(axis=0)

# 3. Apply Condition (ii): < 90% in at least one ST
cond_ii = (prevalence_by_st < 0.90).any(axis=0)

# 4. Combine conditions (must satisfy BOTH)
valid_features = cond_i & cond_ii

# 5. Extract the list of columns that passed the filter
selected_cols = valid_features[valid_features].index.tolist()

# 6. Create the final dataframe with reordering: [Isolate ID] + [Features] + [MLST]
final_df_list = ['Isolate ID'] + selected_cols + ['MLST']

# Concatenate and reorder
final_filtered_df = pd.concat([metadata[['Isolate ID']], features[selected_cols], metadata[['MLST']]], axis=1)
final_filtered_df = final_filtered_df[final_df_list]

# Save the filtered result
final_filtered_df.to_csv(output_path, index=False)

print(f"--- Filtering Complete ---")
print(f"Original feature count: {features.shape[1]}")
print(f"Features after 40%/90% filter: {len(selected_cols)}")
print(f"Resultant Dataset Shape: {final_filtered_df.shape}")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from statsmodels.stats.multitest import multipletests

# 1. Load your dataset
file_path = r'D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Filtered_ML_Ready_Dataset.csv'
df = pd.read_csv(file_path)

# 2. Identify Metadata and Features
# 'Isolate ID' is metadata, 'MLST' is target, rest are features
target_col = 'MLST'
metadata_col = 'Isolate ID'
feature_cols = [col for col in df.columns if col not in [target_col, metadata_col]]

# 3. Perform Statistical Tests
results = []
print("Running Chi-Square tests...")

for feature in feature_cols:
    contingency_table = pd.crosstab(df[target_col], df[feature])
    
    if contingency_table.shape[0] > 1 and contingency_table.shape[1] > 1:
        chi2, p, dof, ex = chi2_contingency(contingency_table, correction=False)
        results.append({'feature': feature, 'p_value': p})
    else:
        results.append({'feature': feature, 'p_value': np.nan})

results_df = pd.DataFrame(results).dropna()

# 4. Multiple Testing Correction
results_df['p_adj'] = multipletests(results_df['p_value'], method='fdr_bh')[1]

# 5. Filter for significant features
significant_df = results_df[results_df['p_adj'] < 0.005]
significant_feature_names = significant_df['feature'].tolist()

print(f"Number of significant features identified: {len(significant_feature_names)}")

# 6. Create and Save the Filtered Dataset
# Structure: [Isolate ID] + [Significant Features] + [MLST]
cols_to_keep = [metadata_col] + significant_feature_names + [target_col]
final_filtered_df = df[cols_to_keep]

# Save the actual dataset
output_path = r'D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Statistically_Significant_Dataset.csv'
final_filtered_df.to_csv(output_path, index=False)

print(f"Filtered dataset saved successfully to: {output_path}")
print(f"Final dataset shape: {final_filtered_df.shape}")
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
file_path = r'D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Statistically_Significant_Dataset.csv'
data = pd.read_csv(file_path)

# 2. Set Isolate ID as index and separate data
data.set_index(data.columns[0], inplace=True)
mlst_data = data['MLST']
gene_data = data.drop(columns=['MLST'])

# 3. Scale the data (Standardization is crucial for PCA)
scaler = StandardScaler()
gene_data_scaled = scaler.fit_transform(gene_data)

# 4. Perform PCA
pca = PCA(n_components=10)
pca_result = pca.fit_transform(gene_data_scaled)

# 5. Create DataFrame for results
pca_df = pd.DataFrame(data=pca_result[:, :2], columns=['PC1', 'PC2'], index=data.index)
pca_df['MLST'] = mlst_data

# Calculate centroids
centroids = pca_df.groupby('MLST')[['PC1', 'PC2']].mean()

# 6. Plotting
sns.set(style='ticks')
fig, ax = plt.subplots(figsize=(10, 10), dpi=300)

# Colors
colors = dict(ST131='firebrick', ST1193='darkorange', ST69='aquamarine', ST73='darkgreen',
              ST38='gold', ST10='darkgoldenrod', ST95='plum', ST12='slateblue',
              ST648='chocolate', ST410='lightcoral', ST127='greenyellow', ST405='steelblue',
              ST167='orangered', ST617='teal', ST58='hotpink', ST101='darkred',
              ST44='mediumseagreen')

sns.scatterplot(x='PC1', y='PC2', hue='MLST', palette=colors, data=pca_df, s=100, alpha=0.8, ax=ax)

# Add centroid labels
for mlst, (x, y) in centroids.iterrows():
    ax.text(x, y, mlst, fontsize=10, weight='bold', ha='center', 
            color='black', bbox=dict(facecolor='white', alpha=0.6, edgecolor='none'))

ax.set_title('Principal Component Analysis (PCA)', fontsize=16)
ax.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)', fontsize=14)
ax.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)', fontsize=14)
ax.grid(False)

# 7. Cumulative Scree Plot Inset
inset_ax = fig.add_axes([0.65, 0.65, 0.2, 0.2])
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)
inset_ax.plot(range(1, 11), cumulative_variance, marker='o', linestyle='-', color='black')
inset_ax.set_title('Cumulative Variance', fontsize=9)
inset_ax.set_ylabel('Variance (%)', fontsize=8)
inset_ax.set_xticks(range(1, 11))
inset_ax.tick_params(axis='both', which='major', labelsize=7)

plt.savefig('PCA_Analysis.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.spatial import ConvexHull

# 1. Load Data
file_path = r'D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Statistically_Significant_Dataset.csv'
data = pd.read_csv(file_path)
data.set_index(data.columns[0], inplace=True)
mlst_data = data['MLST']
gene_data = data.drop(columns=['MLST'])

# 2. PCA
scaler = StandardScaler()
gene_data_scaled = scaler.fit_transform(gene_data)
pca = PCA(n_components=10)
pca_result = pca.fit_transform(gene_data_scaled)
pca_df = pd.DataFrame(data=pca_result[:, :2], columns=['PC1', 'PC2'], index=data.index)
pca_df['MLST'] = mlst_data

# 3. Plot Setup
fig, ax = plt.subplots(figsize=(10, 8), dpi=300)
colors = dict(ST131='firebrick', ST1193='darkorange', ST69='aquamarine', ST73='darkgreen',
              ST38='gold', ST10='darkgoldenrod', ST95='plum', ST12='slateblue',
              ST648='chocolate', ST410='lightcoral', ST127='greenyellow', ST405='steelblue',
              ST167='orangered', ST617='teal', ST58='hotpink', ST101='darkred',
              ST44='mediumseagreen')

# 4. Scatter Plot (Reduced Alpha and Size for Clarity)
sns.scatterplot(x='PC1', y='PC2', hue='MLST', palette=colors, data=pca_df, 
                s=20, alpha=0.5, edgecolor=None, ax=ax, legend=False)

# 5. Add Convex Hulls (Clean boundaries for each ST)
for st in pca_df['MLST'].unique():
    points = pca_df[pca_df['MLST'] == st][['PC1', 'PC2']].values
    if len(points) >= 3:
        hull = ConvexHull(points)
        for simplex in hull.simplices:
            plt.plot(points[simplex, 0], points[simplex, 1], color=colors.get(st), lw=1, alpha=0.5)
        # Fill the hull
        plt.fill(points[hull.vertices, 0], points[hull.vertices, 1], color=colors.get(st), alpha=0.1)

# 6. Aesthetic Cleanup
sns.despine()
ax.set_title('UPEC Lineage Genomic Markers (PCA)', fontsize=14, weight='bold')
ax.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)', fontsize=12)
ax.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)', fontsize=12)

# 7. Inset Plot (Manually placing to ensure no overlap)
# We place it using axes coordinates [left, bottom, width, height]
inset_ax = fig.add_axes([0.15, 0.65, 0.15, 0.15]) 
cum_var = np.cumsum(pca.explained_variance_ratio_) * 100
inset_ax.plot(range(1, 11), cum_var, marker='o', markersize=3, color='black', lw=1)
inset_ax.set_title('Cumulative Variance (%)', fontsize=7)
inset_ax.tick_params(labelsize=6)
sns.despine(ax=inset_ax)

# 8. Save
plt.savefig('Final_Clean_PCA.svg', format='svg', bbox_inches='tight')
plt.savefig('Final_Clean_PCA.tiff', dpi=300, format='tiff', bbox_inches='tight')
plt.show()
```
