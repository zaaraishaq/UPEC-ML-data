```python
import os
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize
from sklearn.metrics import (
    accuracy_score, f1_score, recall_score, precision_score,
    classification_report, confusion_matrix, roc_curve, auc, roc_auc_score,
    log_loss
)
# Oversampling for handling class imbalance in binary/categorical datasets
from imblearn.over_sampling import SMOTEN

# Models
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier
from sklearn.calibration import CalibratedClassifierCV
from sklearn.neural_network import MLPClassifier

try:
    from IPython.display import display
except ImportError:
    def display(df):
        print(df.to_string())

# Suppress runtime, future, and optimization warnings
warnings.filterwarnings('ignore', category=UserWarning, module='sklearn')
warnings.filterwarnings('ignore', category=FutureWarning, module='seaborn')
warnings.filterwarnings('ignore', category=RuntimeWarning, module='numpy')

# ==========================================
# CONFIGURATION
# ==========================================
DATA_PATH = r"D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Statistically_Significant_Dataset.csv"
ID_COLUMN = "Isolate ID"    # Identifier column (e.g., 'CMZE092')
TARGET_COLUMN = "MLST"      # Target MLST sequence types

print("Starting pipeline: Multi-Model Training and Evaluation with SMOTE-N...")
print("-" * 60)

# ==========================================
# 1. LOAD & PREPARE DATA
# ==========================================
try:
    print(f"Using dataset from: {DATA_PATH}")
    if not os.path.exists(DATA_PATH):
        raise FileNotFoundError(f"File not found at: {DATA_PATH}")
    
    df = pd.read_csv(DATA_PATH)
    print(f"Data loaded successfully with {df.shape[0]} samples and {df.shape[1]} features")
except Exception as e:
    print(f"Error loading file: {e}")
    exit()

# Verify baseline columns
if TARGET_COLUMN not in df.columns:
    print(f"Error: Target column '{TARGET_COLUMN}' not found in dataset.")
    exit()

if ID_COLUMN not in df.columns:
    print(f"Error: ID column '{ID_COLUMN}' not found in dataset.")
    exit()

# Separate features and target
X = df.drop(columns=[TARGET_COLUMN])
y = df[TARGET_COLUMN]

# Fix target labels format safely (Ensure clean string formats such as "ST10")
y = y.apply(lambda x: f"ST{int(x)}" if pd.api.types.is_numeric_dtype(type(x)) and not pd.isna(x) else str(x))

# Move 'Isolate ID' to index so features are strictly numeric
X = X.set_index(ID_COLUMN)
print(f"Successfully moved '{ID_COLUMN}' to dataframe index to prevent scaling errors.")

# Impute missing feature values if any exist
if X.isnull().values.any():
    print("Warning: Missing values found in features. Performing automated imputation.")
    for col in X.columns:
        if X[col].isnull().any():
            if pd.api.types.is_numeric_dtype(X[col]):
                X[col].fillna(X[col].mean(), inplace=True)
            else:
                X[col].fillna(X[col].mode()[0], inplace=True)

# Drop lingering non-numeric columns if any exist in features
text_cols = X.select_dtypes(include=['object', 'category']).columns
if len(text_cols) > 0:
    print(f"Warning: Dropping remaining non-numeric columns from features: {list(text_cols)}")
    X = X.drop(columns=text_cols)

# ==========================================
# 2. ENCODE TARGET CLASSES
# ==========================================
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
classes = label_encoder.classes_
print(f"Encoded {len(classes)} unique classes: {classes}")

# Catch tiny target sub-populations (< 2 instances) that break stratified splitting 
class_counts = pd.Series(y_encoded).value_counts()
single_sample_classes = class_counts[class_counts < 2].index.tolist()
if single_sample_classes:
    single_sample_names = label_encoder.inverse_transform(single_sample_classes).tolist()
    print(f"Warning: Classes with < 2 samples identified: {single_sample_names}. Reverting to fallback split strategy.")

# ==========================================
# 3. DATA SPLITTING, RESAMPLING & SCALING
# ==========================================
try:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
    )
    print("Data split successfully using stratification.")
except ValueError as e:
    print(f"Stratification failed: {e}. Reverting split strategy to standard fallback mode.")
    X_train, X_test, y_train, y_test = train_test_split(
        X, y_encoded, test_size=0.2, random_state=42
    )
    print("Warning: Split performed without stratification.")

# Apply SMOTE-N for categorical/binary features to handle class imbalance
X_train_resampled, y_train_resampled = X_train.copy(), y_train.copy()
try:
    # k_neighbors=1 acts as a safeguard when extremely rare minority classes are present
    smoten = SMOTEN(random_state=42, k_neighbors=1)
    X_train_resampled, y_train_resampled = smoten.fit_resample(X_train, y_train)
    print(f"SMOTE-N Over-sampling Complete. Original train configuration: {X_train.shape}, Balanced Resampled shape: {X_train_resampled.shape}")
except Exception as e:
    print(f"Warning: SMOTE-N could not run (likely due to classes with insufficient instances): {e}. Proceeding with original distributions.")

# Build scaled matrix instances for distance & coordinate-dependent architectures
try:
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train_resampled)
    X_test_scaled = scaler.transform(X_test)
    print("Feature scaling transformations applied successfully.")
except Exception as e:
    print(f"Error establishing scaling parameters: {e}")
    exit()

# ==========================================
# 4. INITIALIZE MODELS SETUP
# ==========================================
models = {
    "Random Forest": RandomForestClassifier(
        class_weight='balanced', n_estimators=150, max_depth=15, random_state=42, n_jobs=-1
    ),
    "Decision Tree": DecisionTreeClassifier(
        class_weight='balanced', max_depth=7, random_state=42
    ),
    "KNN": KNeighborsClassifier(
        n_neighbors=7, weights='distance', n_jobs=-1
    ),
    "SVM (Linear)": CalibratedClassifierCV(
        SVC(kernel='linear', class_weight='balanced', probability=True, random_state=42, max_iter=15000),
        cv=3
    ),
    "SVM (RBF)": CalibratedClassifierCV(
        SVC(kernel='rbf', class_weight='balanced', probability=True, random_state=42, max_iter=15000),
        cv=3
    ),
    "GaussianNB": GaussianNB(),
    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=150, random_state=42
    ),
    "XGBoost": XGBClassifier(
        eval_metric='mlogloss', random_state=42, n_estimators=150, learning_rate=0.1,
        n_jobs=-1
    ),
    "AdaBoost": AdaBoostClassifier(
        n_estimators=150, random_state=42
    ),
    "MLP Classifier": MLPClassifier(
        hidden_layer_sizes=(128, 64), max_iter=1000, alpha=0.0001, solver='adam',
        random_state=42, learning_rate_init=.001, early_stopping=True, verbose=False
    )
}

# Distance metric, hyperplanes, or neural nodes depend explicitly on standardized inputs
scale_sensitive = ["KNN", "SVM (Linear)", "SVM (RBF)", "MLP Classifier"]

# ==========================================
# 5. VISUALIZATION AUXILIARIES
# ==========================================
def plot_learning_curve(model, X_train, y_train, model_name):
    try:
        if hasattr(model, 'evals_result_'):
            results = model.evals_result()
            plt.figure(figsize=(8, 4))
            if 'validation_0' in results and 'mlogloss' in results['validation_0']:
                plt.plot(results['validation_0']['mlogloss'], label='Train Loss')
            if 'validation_1' in results and 'mlogloss' in results['validation_1']:
                plt.plot(results['validation_1']['mlogloss'], label='Validation Loss')
            plt.title(f'{model_name} - Learning Curve')
            plt.xlabel('Boosting Round')
            plt.ylabel('Loss')
            plt.legend()
            plt.show()
        elif hasattr(model, 'train_score_'):
            plt.figure(figsize=(8, 4))
            plt.plot(np.arange(len(model.train_score_)), model.train_score_, label='Train Deviance')
            plt.title(f'{model_name} - Learning Curve')
            plt.xlabel('Boosting Round')
            plt.ylabel('Deviance')
            plt.legend()
            plt.show()
        elif isinstance(model, MLPClassifier) and hasattr(model, 'loss_curve_'):
            plt.figure(figsize=(8, 4))
            plt.plot(model.loss_curve_, label='Train Loss')
            plt.title(f'{model_name} - Learning Curve (Loss)')
            plt.xlabel('Epoch')
            plt.ylabel('Loss')
            plt.legend()
            plt.show()
    except Exception as e:
        print(f"Could not plot learning curve for {model_name}: {str(e)}")

def plot_roc_auc(y_test, y_prob, classes, model_name, all_roc_data=None):
    plt.figure(figsize=(10, 8))
    n_classes = len(classes)
    y_test_binarized = label_binarize(y_test, classes=np.arange(n_classes))
    
    if y_prob is None or y_prob.shape[1] != n_classes:
        if y_prob is not None and y_test_binarized.size > 0 and y_prob.size > 0:
            fpr_micro, tpr_micro, _ = roc_curve(y_test_binarized.ravel(), y_prob.ravel())
            roc_auc_micro = auc(fpr_micro, tpr_micro)
            plt.plot(fpr_micro, tpr_micro, label=f'{model_name} micro-average (AUC = {roc_auc_micro:.2f})', linestyle=':', linewidth=2)
        plt.plot([0, 1], [0, 1], 'k--', lw=2)
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'ROC Curve for {model_name} (Micro-average only)')
        plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
        plt.show()
        return

    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    plottable_classes = []
    
    for i in range(n_classes):
        if len(np.unique(y_test_binarized[:, i])) > 1:
            fpr[i], tpr[i], _ = roc_curve(y_test_binarized[:, i], y_prob[:, i])
            roc_auc[i] = auc(fpr[i], tpr[i])
            plottable_classes.append(i)
            
    fpr["micro"], tpr["micro"], _ = roc_curve(y_test_binarized.ravel(), y_prob.ravel())
    roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])
    
    colors = plt.cm.rainbow(np.linspace(0, 1, len(plottable_classes)))
    for idx, color in zip(plottable_classes, colors):
        plt.plot(fpr[idx], tpr[idx], color=color, lw=2, label=f'Class {classes[idx]} (AUC = {roc_auc[idx]:.2f})')
        
    plt.plot(fpr["micro"], tpr["micro"], label=f'Micro-average (AUC = {roc_auc["micro"]:.2f})', color='deeppink', linestyle=':', linewidth=4)
    plt.plot([0, 1], [0, 1], 'k--', lw=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'ROC Curves for {model_name}')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
    
    if all_roc_data is not None:
        all_roc_data[model_name] = {"fpr_micro": fpr["micro"], "tpr_micro": tpr["micro"], "roc_auc_micro": roc_auc["micro"]}

# ==========================================
# 6. ENSEMBLE TRAINING & BENCHMARK LOOP
# ==========================================
results = []
all_roc_data = {}

for name, model in models.items():
    print(f"\n{'='*60}")
    print(f"Training and Evaluating Model: {name}")
    print(f"{'='*60}")
    
    # Direct scaled vs raw data frames to models based on scale sensitivity constraints
    if name in scale_sensitive:
        train_data = X_train_scaled
        test_data = X_test_scaled
    else:
        train_data = X_train_resampled.values  # Convert to array to prevent header warning calls
        test_data = X_test.values

    # Fit Model
    start_time = time.time()
    if name == "XGBoost":
        model.fit(train_data, y_train_resampled, eval_set=[(train_data, y_train_resampled), (test_data, y_test)], verbose=False)
    else:
        model.fit(train_data, y_train_resampled)
    training_time = time.time() - start_time
    
    # Plot Convergence Profiles
    plot_learning_curve(model, train_data, y_train_resampled, name)
    
    # Predictions
    y_pred = model.predict(test_data)
    y_prob = model.predict_proba(test_data) if hasattr(model, "predict_proba") else None
    
    # Evaluate Validation Set Performance
    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
    loss = log_loss(y_test, y_prob) if y_prob is not None else np.nan
    
    roc_auc = np.nan
    if y_prob is not None:
        y_test_binarized = label_binarize(y_test, classes=np.arange(len(classes)))
        if y_test_binarized.size > 0 and y_prob.size > 0 and y_test_binarized.shape[1] == y_prob.shape[1] and np.sum(y_test_binarized) > 0:
            roc_auc = roc_auc_score(y_test_binarized, y_prob, multi_class='ovr', average='weighted')
            plot_roc_auc(y_test, y_prob, classes, name, all_roc_data)
        else:
            if y_test_binarized.size > 0 and y_prob.size > 0:
                plot_roc_auc(y_test, y_prob, classes, name, all_roc_data)

    results.append({
        'Model': name, 'Accuracy': accuracy, 'F1 Score': f1, 'Recall': recall,
        'Precision': precision, 'ROC AUC': roc_auc, 'Log Loss': loss, 'Training Time (s)': training_time
    })
    
    print(f"\n{name} Analysis Summary:")
    print(f"- Process/Train Execution window: {training_time:.3f}s")
    print(f"- Overall Accuracy: {accuracy:.4f}")
    print(f"- Weighted F1 Metric: {f1:.4f}")
    print(f"- ROC AUC Average: {roc_auc if not np.isnan(roc_auc) else 'N/A'}")
    print(f"- Logarithmic Loss Metric: {loss if not np.isnan(loss) else 'N/A'}")
    
    print("\nClassification Report View:")
    print(classification_report(y_test, y_pred, target_names=classes, zero_division=0))
    
    # Heatmap Confusion Matrix
    plt.figure(figsize=(8, 6))
    all_possible_labels = np.arange(len(classes))
    cm = confusion_matrix(y_test, y_pred, labels=all_possible_labels)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
    plt.title(f'Confusion Matrix Map - {name}')
    plt.xlabel('Predicted MLST Grouping')
    plt.ylabel('Actual MLST Grouping')
    plt.tight_layout()
    plt.show()

# ==========================================
# 7. EXPORT DATA SUMMARY & FINAL GRAPH
# ==========================================
if all_roc_data:
    plt.figure(figsize=(10, 8))
    for model_name, data in all_roc_data.items():
        plt.plot(data["fpr_micro"], data["tpr_micro"], label=f'{model_name} (Micro AUC = {data["roc_auc_micro"]:.2f})')
    plt.plot([0, 1], [0, 1], 'k--', lw=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Combined Performance ROC Analysis (Micro-average comparison)')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

print("\n" + "="*60)
print("COMPREHENSIVE RUN METRICS COMPARISON DATAFRAME:")
print("="*60)
results_df = pd.DataFrame(results)
display(results_df)

try:
    results_df.to_csv('model_performance_results.csv', index=False)
    print("\nMetrics matrix saved successfully to 'model_performance_results.csv'")
except Exception as e:
    print(f"Error preserving performance metrics spreadsheet: {e}")

print("\nModel training and validation pipeline successfully finalized!")
```

```python
import os
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Framework modules
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize
from sklearn.metrics import roc_auc_score
# High-speed oversampler perfectly suited for high-dimensional data
from imblearn.over_sampling import RandomOverSampler 

# Machine learning models
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier
from sklearn.calibration import CalibratedClassifierCV
from sklearn.neural_network import MLPClassifier

try:
    from IPython.display import display
except ImportError:
    def display(df):
        print(df.to_string())

warnings.filterwarnings('ignore', category=UserWarning, module='sklearn')
warnings.filterwarnings('ignore', category=FutureWarning, module='seaborn')
warnings.filterwarnings('ignore', category=RuntimeWarning, module='numpy')

# ==========================================
# CONFIGURATION
# ==========================================
DATA_PATH = r"D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Statistically_Significant_Dataset.csv"
ID_COLUMN = "Isolate ID"    
TARGET_COLUMN = "MLST"      
NUM_FOLDS = 5

print(f"Initializing High-Speed {NUM_FOLDS}-Fold Stratified Cross-Validation Pipeline...")
print("-" * 75)

# ==========================================
# 1. LOAD & PREPARE DATA
# ==========================================
try:
    if not os.path.exists(DATA_PATH):
        raise FileNotFoundError(f"File not found at: {DATA_PATH}")
    df = pd.read_csv(DATA_PATH)
    print(f"Dataset successfully integrated: {df.shape[0]} strains, {df.shape[1]} features.")
except Exception as e:
    print(f"Fatal load failure: {e}")
    exit()

X = df.drop(columns=[TARGET_COLUMN])
y = df[TARGET_COLUMN]

y = y.apply(lambda x: f"ST{int(x)}" if pd.api.types.is_numeric_dtype(type(x)) and not pd.isna(x) else str(x))
X = X.set_index(ID_COLUMN)

text_cols = X.select_dtypes(include=['object', 'category']).columns
if len(text_cols) > 0:
    X = X.drop(columns=text_cols)

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
classes = label_encoder.classes_
n_classes = len(classes)

# ==========================================
# 2. DEFINE SYSTEM MODELS (HIGH-SPEED GENOMIC OPTIMIZED)
# ==========================================
def initialize_models():
    return {
        "Random Forest": RandomForestClassifier(class_weight='balanced', n_estimators=100, max_depth=12, random_state=42, n_jobs=-1),
        "Decision Tree": DecisionTreeClassifier(class_weight='balanced', max_depth=15, random_state=42),
        "Gradient Boosting": GradientBoostingClassifier(n_estimators=80, max_depth=3, random_state=42),
        "XGBoost": XGBClassifier(eval_metric='mlogloss', random_state=42, n_estimators=80, learning_rate=0.1, n_jobs=-1),
        "AdaBoost": AdaBoostClassifier(estimator=DecisionTreeClassifier(max_depth=3, random_state=42), n_estimators=80, random_state=42),
        "SVM (RBF)": CalibratedClassifierCV(SVC(kernel='rbf', class_weight='balanced', probability=True, random_state=42, max_iter=1000), cv=3),
        "SVM (Linear)": CalibratedClassifierCV(SVC(kernel='linear', class_weight='balanced', probability=True, random_state=42, max_iter=1000), cv=3),
        "KNN": KNeighborsClassifier(n_neighbors=5, weights='distance', n_jobs=-1),
        "GaussianNB": GaussianNB(),
        "MLP Classifier": MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=200, alpha=0.001, solver='adam', random_state=42, early_stopping=True)
    }

scale_sensitive = ["KNN", "SVM (Linear)", "SVM (RBF)", "MLP Classifier"]
model_execution_order = list(initialize_models().keys())

final_table_data = []

# ==========================================
# 3. CORE STRATIFIED CROSS-VALIDATION LOOP
# ==========================================
for idx, model_name in enumerate(model_execution_order, 1):
    start_model_time = time.time()
    print(f"\n[{idx}/{len(model_execution_order)}] Running 5-Fold Stratified CV for: {model_name}")
    print("." * 60)
    
    fold_roc_aucs = []
    all_fold_loss_curves = []
    
    skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=42)
    
    for fold_idx, (train_idx, val_idx) in enumerate(skf.split(X, y_encoded)):
        X_train_fold, X_val_fold = X.iloc[train_idx].copy(), X.iloc[val_idx].copy()
        y_train_fold, y_val_fold = y_encoded[train_idx], y_encoded[val_idx]
        
        # High-Speed Random Over-Sampling to balance dataset instantly
        try:
            ros = RandomOverSampler(random_state=42)
            X_train_res, y_train_res = ros.fit_resample(X_train_fold, y_train_fold)
        except Exception as e:
            X_train_res, y_train_res = X_train_fold.copy(), y_train_fold.copy()
            
        if model_name in scale_sensitive:
            scaler = StandardScaler()
            X_train_input = scaler.fit_transform(X_train_res)
            X_val_input = scaler.transform(X_val_fold)
        else:
            X_train_input = X_train_res.values if hasattr(X_train_res, 'values') else X_train_res
            X_val_input = X_val_fold.values
            
        current_model = initialize_models()[model_name]
        
        if model_name == "XGBoost":
            current_model.fit(X_train_input, y_train_res, eval_set=[(X_train_input, y_train_res), (X_val_input, y_val_fold)], verbose=False)
            evals_res = current_model.evals_result()
            if 'validation_1' in evals_res and 'mlogloss' in evals_res['validation_1']:
                all_fold_loss_curves.append(evals_res['validation_1']['mlogloss'])
        elif model_name == "MLP Classifier":
            current_model.fit(X_train_input, y_train_res)
            if hasattr(current_model, 'loss_curve_'):
                all_fold_loss_curves.append(current_model.loss_curve_)
        else:
            current_model.fit(X_train_input, y_train_res)
            
        if hasattr(current_model, "predict_proba"):
            probabilities = current_model.predict_proba(X_val_input)
            y_val_binarized = label_binarize(y_val_fold, classes=np.arange(n_classes))
            
            try:
                auc_score = roc_auc_score(y_val_binarized, probabilities, multi_class='ovr', average='weighted')
                fold_roc_aucs.append(auc_score)
            except ValueError:
                auc_score = roc_auc_score(y_val_fold, probabilities, multi_class='ovr', average='weighted')
                fold_roc_aucs.append(auc_score)
        else:
            fold_roc_aucs.append(np.nan)

    # ==========================================
    # 4. GENERATE INDIVIDUAL LEARNING CURVES
    # ==========================================
    if all_fold_loss_curves:
        plt.figure(figsize=(7, 3.5))
        min_length = min(len(curve) for curve in all_fold_loss_curves)
        truncated_curves = [curve[:min_length] for curve in all_fold_loss_curves]
        
        mean_loss = np.mean(truncated_curves, axis=0)
        std_loss = np.std(truncated_curves, axis=0)
        epochs = np.arange(1, min_length + 1)
        
        plt.plot(epochs, mean_loss, color='blue', lw=2, label='Mean Validation Loss')
        plt.fill_between(epochs, mean_loss - std_loss, mean_loss + std_loss, color='blue', alpha=0.15, label='$\pm$1 SD Variance')
        plt.title(f'{model_name} Cross-Validation Learning Curve')
        plt.xlabel('Iteration / Training Epoch')
        plt.ylabel('Validation Loss')
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.legend()
        plt.tight_layout()
        plt.show()

    mean_auc = np.mean(fold_roc_aucs)
    std_auc = np.std(fold_roc_aucs)
    print(f"Finished {model_name} in {time.time() - start_model_time:.2f}s | Mean ROC AUC: {mean_auc:.4f}")
    
    final_table_data.append({
        'No.': int(idx),
        'Model': model_name,
        'Mean ROC AUC': round(mean_auc, 9),
        'Std ROC AUC': round(std_auc, 9),
        'Scores': [round(score, 8) for score in fold_roc_aucs]
    })

# ==========================================
# 5. RENDER FINAL MANUSCRIPT FORMAT
# ==========================================
print("\n" + "=" * 85)
print("Table 2. 5-fold stratified cross-validation performance evaluation of machine learning models.")
print("=" * 85)

table_2_df = pd.DataFrame(final_table_data).set_index('No.')
display(table_2_df)

output_filename = 'Table_2_Stratified_CV_ROC_AUC_Results.csv'
table_2_df.to_csv(output_filename, index=True)
print(f"\nMatrix calculations written successfully to disk location: '{output_filename}'")
```

```python
import os
import warnings
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import RFE
from imblearn.over_sampling import RandomOverSampler
import shap

# Force immediate flushing of logs and suppress library warnings
warnings.filterwarnings('ignore')
os.environ['PYTHONUNBUFFERED'] = '1'

# ==========================================
# CONFIGURATION & DATA RECOVERY
# ==========================================
DATA_PATH = r"D:\UPEC PhD Project\UPEC_ML_Data\Frontiers_Word_Templates\Revision 2.0\Analysis\Datasets\Statistically_Significant_Dataset.csv"
ID_COLUMN = "Isolate ID"    
TARGET_COLUMN = "MLST"      
RANDOM_STATE = 42

print("Starting Phase 1: Train/Test Split, RFE, & Test-Set SHAP Computation...", flush=True)
print("-" * 75, flush=True)

# 1. Load Data
df = pd.read_csv(DATA_PATH)
X_raw = df.drop(columns=[TARGET_COLUMN]).set_index(ID_COLUMN)
text_cols = X_raw.select_dtypes(include=['object', 'category']).columns
if len(text_cols) > 0:
    X_raw = X_raw.drop(columns=text_cols)

y_formatted = df[TARGET_COLUMN].apply(lambda x: f"ST{int(x)}" if pd.api.types.is_numeric_dtype(type(x)) and not pd.isna(x) else str(x))
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y_formatted)

# 2. SPLIT DATA FIRST (80% Train, 20% Test) to prevent data leakage
print("Splitting data into Training and Unseen Test sets...", flush=True)
X_train, X_test, y_train, y_test = train_test_split(
    X_raw, y_encoded, test_size=0.20, random_state=RANDOM_STATE, stratify=y_encoded
)

# 3. Balance ONLY the Training Dataset
print("Balancing training data...", flush=True)
ros = RandomOverSampler(random_state=RANDOM_STATE)
X_train_balanced_np, y_train_balanced = ros.fit_resample(X_train.values, y_train)
X_train_balanced_df = pd.DataFrame(X_train_balanced_np, columns=X_raw.columns)

# 4. Two-Tiered Feature Selection on Training Data
print("Identifying top markers using ONLY training data...", flush=True)
initial_model = RandomForestClassifier(
    class_weight='balanced', n_estimators=100, max_depth=12, random_state=RANDOM_STATE, n_jobs=1
)
initial_model.fit(X_train_balanced_df, y_train_balanced)

# Tier 1: Gini Impurity Screening (>= 0.001)
importances = initial_model.feature_importances_
gini_mask = importances >= 0.001
features_passing_gini = X_raw.columns[gini_mask].tolist()

print(f"\n-> Tier 1 Complete: {len(features_passing_gini)} features passed the >=0.001 Gini threshold.")

# Filter training data to only include these pre-screened features
X_train_tier1 = X_train_balanced_df[features_passing_gini]

# Tier 2: Recursive Feature Elimination (RFE) to isolate the top 20
print("-> Running RFE to prune down to the top 20 features (this may take a moment)...", flush=True)
rfe = RFE(estimator=initial_model, n_features_to_select=20, step=0.05)
rfe.fit(X_train_tier1, y_train_balanced)

top_20_features = X_train_tier1.columns[rfe.support_].tolist()

print("\n-> Top 20 Features isolated via RFE:")
print(top_20_features)

# Internal Validation: 5-Fold Cross-Validation on the top 20 features
print("\n-> Validating predictive stability of the top 20 features via 5-Fold CV...", flush=True)
cv_scores = cross_val_score(
    initial_model, X_train_tier1[top_20_features], y_train_balanced, cv=5, scoring='accuracy'
)
print(f"   5-Fold CV Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std() * 2:.4f})")

# 5. Filter datasets to only include these top 20 features
X_train_top20 = X_train_balanced_df[top_20_features]
X_test_top20 = X_test[top_20_features] # The raw, unbalanced, unseen test data!

# 6. Train Final Champion Model
print("\nTraining final champion model...", flush=True)
final_model = RandomForestClassifier(
    class_weight='balanced', n_estimators=100, max_depth=12, random_state=RANDOM_STATE, n_jobs=1
)
final_model.fit(X_train_top20, y_train_balanced)

# 7. Generate SHAP Values ON THE UNSEEN TEST SET
print("Extracting SHAP matrices strictly from the Unseen Test Set...", flush=True)
explainer = shap.TreeExplainer(final_model)

# Notice we are passing X_test_top20 here, not the whole dataset!
shap_values = explainer.shap_values(X_test_top20.values, check_additivity=False)

# Resolve multi-class SHAP shapes
if isinstance(shap_values, list):
    mean_abs_shap_per_class = np.array([np.mean(np.abs(class_shap), axis=0) for class_shap in shap_values])
else:
    if len(shap_values.shape) == 3:  
        mean_abs_shap_per_class = np.mean(np.abs(shap_values), axis=0).T
    else:
        mean_abs_shap_per_class = np.atleast_2d(np.mean(np.abs(shap_values), axis=0))

heatmap_df = pd.DataFrame(
    mean_abs_shap_per_class, 
    index=label_encoder.classes_, 
    columns=top_20_features
)

# Crucial update for Phase 2: We need to pass the TEST variables so Phase 2 plots the right data
X_top20_raw = X_test_top20 # Overwrite this so Phase 2 uses the test data for plotting
y_encoded = y_test         # Overwrite this so Phase 2 uses the test labels for prevalence/violins

print("\n✅ Phase 1 Complete! Memory buffered. You can now run Cell 2.", flush=True)
```

```python
import matplotlib
# 🛑 CRASH PREVENTION: FORCE BACKGROUND RENDERING
matplotlib.use('Agg') 
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import gc
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, roc_curve, auc
)
from sklearn.preprocessing import label_binarize

# Set clean, publication aesthetics safely
plt.style.use('default')
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 13,
    'figure.titlesize': 14
})

print("\nStarting Phase 3: Model Performance & Evaluation...", flush=True)
print("-" * 75, flush=True)

save_formats = ['png', 'svg', 'tiff']

# 1. Generate Predictions on the UNSEEN Test Set
print("Generating predictions on the test set...", flush=True)
y_pred = final_model.predict(X_test_top20)
y_prob = final_model.predict_proba(X_test_top20)

# 2. Calculate Global Metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)

print(f"\nGlobal Test Set Metrics:")
print(f" -> Accuracy:  {accuracy:.4f}")
print(f" -> Precision: {precision:.4f} (Weighted)")
print(f" -> Recall:    {recall:.4f} (Weighted)")
print(f" -> F1-Score:  {f1:.4f} (Weighted)")

# 3. Export Detailed Classification Report (CSV)
print("\nExporting detailed classification report...", flush=True)
report_dict = classification_report(y_test, y_pred, target_names=label_encoder.classes_, output_dict=True, zero_division=0)
report_df = pd.DataFrame(report_dict).transpose()
report_csv_path = 'Table_Classification_Report_Metrics.csv'
report_df.to_csv(report_csv_path)
print(f" -> Saved to {report_csv_path}")

# ==========================================
# FIGURE 6: CONFUSION MATRIX
# ==========================================
print("\nGenerating High-Contrast Confusion Matrix...", flush=True)
try:
    cm = confusion_matrix(y_test, y_pred)
    
    fig, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(
        cm, annot=True, fmt='d', cmap='Blues', 
        xticklabels=label_encoder.classes_, 
        yticklabels=label_encoder.classes_,
        cbar_kws={'label': 'Number of Isolates'},
        ax=ax
    )
    
    plt.setp(ax.get_xticklabels(), rotation=45, horizontalalignment='right', fontweight='bold')
    plt.setp(ax.get_yticklabels(), rotation=0, fontweight='bold')
    
    ax.set_title("Test Set Confusion Matrix", fontweight='bold', pad=15)
    ax.set_ylabel("Actual MLST Lineage", fontweight='bold')
    ax.set_xlabel("Predicted MLST Lineage", fontweight='bold')
    
    fig.tight_layout()
    for ext in save_formats:
        fig.savefig(f'Figure6_Confusion_Matrix.{ext}', dpi=300, bbox_inches='tight')
    plt.close(fig)
    gc.collect()
    print(f" -> Figure 6 Saved Successfully in {save_formats}.", flush=True)
except Exception as e:
    print(f" -> Error in Confusion Matrix: {e}")

# ==========================================
# FIGURE 7: MULTI-CLASS ROC-AUC CURVE
# ==========================================
print("\nGenerating Multi-Class ROC Curves...", flush=True)
try:
    # Binarize the output for One-vs-Rest ROC calculation
    n_classes = len(label_encoder.classes_)
    y_test_bin = label_binarize(y_test, classes=np.arange(n_classes))
    
    fig, ax = plt.subplots(figsize=(11, 8.5))
    
    # Calculate and plot ROC curve for each class
    colors = plt.cm.rainbow(np.linspace(0, 1, n_classes))
    for i, color in zip(range(n_classes), colors):
        # Only plot if the class actually exists in the test split
        if np.sum(y_test_bin[:, i]) > 0:
            fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
            roc_auc = auc(fpr, tpr)
            ax.plot(fpr, tpr, color=color, lw=2, 
                    label=f'{label_encoder.classes_[i]} (AUC = {roc_auc:.2f})')

    # Plot the random guessing diagonal
    ax.plot([0, 1], [0, 1], 'k--', lw=2, label='Random Guessing')
    
    ax.set_xlim([0.0, 1.0])
    ax.set_ylim([0.0, 1.05])
    ax.set_xlabel('False Positive Rate', fontweight='bold')
    ax.set_ylabel('True Positive Rate', fontweight='bold')
    ax.set_title('ROC-AUC Curves per MLST Lineage (Test Set)', fontweight='bold', pad=15)
    
    # Place legend safely outside the plot to avoid overlapping lines
    ax.legend(loc="center left", bbox_to_anchor=(1.05, 0.5), title="Lineages", frameon=True)
    
    fig.tight_layout()
    for ext in save_formats:
        fig.savefig(f'Figure7_ROC_Curves.{ext}', dpi=300, bbox_inches='tight')
    plt.close(fig)
    gc.collect()
    print(f" -> Figure 7 Saved Successfully in {save_formats}.", flush=True)
except Exception as e:
    print(f" -> Error in ROC Curves: {e}")

print("\n✅ Phase 3 Complete! Your ML pipeline is fully finalized.", flush=True)
```
